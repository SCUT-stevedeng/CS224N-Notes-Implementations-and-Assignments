#### RoBERTa
The main idea is to remove Next Sentence Prediction task from BERT, and to pre-train on way larger scale of data.
#### SpanBERT
The main idea is to mask text spans(like phrases, etc.) instead of randomly chosen tokens, and to introduce Span Boundary Objective.
## Encoder-Decoder
Input sequences are splited into a prefix and a suffix. The encoder encodes the entire prefix and the decoder do language modelling autogressively on the suffix.
